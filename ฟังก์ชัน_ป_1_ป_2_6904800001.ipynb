```python
#จงเขียนฟังก์ชันสามตัว คือ is_valid_score(score) ที,คืน True เมื,อคะแนนอยู่ระหว่าง w ถึง vww · to_grade(score)
#ที,แปลงคะแนนเป็นเกรด A (ตั Cงแต่ çw) B (ตั Cงแต่ èw) C (ตั Cงแต่ êw) และ F · print_result(student_id, score)
#ที,แสดงผลในรูปแบบรายงาน โดยถ้าคะแนนไม่ถูกต้องให้แสดงข้อความแจ้งเตือนแทนเกรด ทุกฟังก์ชันต้องมี docstring

def is_valid_score(score):
  return 0 <= score <= 100

def to_grade(score):
  if score >= 80 :
    return "A"
  elif score >= 70 :
    return "B"
  elif score >= 60 :
    return "C"
  else :
    return "F"
student_id = input("รหัสนักศึกษา: ")
score = int(input("คะแนน: "))
print_result(student_id, score)
def print_result(student_id, score):
  if is_valid_score(score):
    print(f"รหัสนักศึกษา {student_id} คะแนน {score} ได้เกรด {to_grade(score)}")
  else:
    print(f"รหัสนักศึกษา {student_id} คะแนน {score} คะแนนไม่ถูกต้อง")
```

```python
#จงเขียนฟังก์ชัน safe_average(total, count) ที,คืนค่าเฉลี,ยปัดทศนิยม U ตําแหน่ง และคืน None เมื,อ count ไม่เป็นบวก
#จากนั Cนเขียนโปรแกรมหลักที,รับค่า PM2.5 จํานวน í วันด้วยตัวแปรห้าตัว แล้วใช้ตัวสะสมหาผลรวมและจํานวน ก่อนเรียก
#safe_average เพื,อแสดงค่าเฉลี,ย โปรแกรมต้องทํางานถูกต้องด้วยเมื,อไม่มีข้อมูลที,ใช้ได้เลย

def safe_average(total, count):
  if count <= 0 :
    return None
  return round(total / count, 2)

total = 0.0
count = 0

PM_1 = float(input("กรุณากรอกค่า PM2.5 Day 1: "))
PM_2 = float(input("กรุณากรอกค่า PM2.5 Day 2: "))
PM_3 = float(input("กรุณากรอกค่า PM2.5 Day 3: "))
PM_4 = float(input("กรุณากรอกค่า PM2.5 Day 4: "))
PM_5 = float(input("กรุณากรอกค่า PM2.5 Day 5: "))

for pm in(PM_1, PM_2, PM_3, PM_4, PM_5):
  total += pm
  count += 1

average = safe_average(total, count)
if average is None :
  print("ไม่มีข้อมูล")
else:
  print("======= ยอดสรุป =======")
  print("Total", total)
  print("Valid Day", count)
  print("Average", average)
  print("======= ยอดสรุป =======")
```
