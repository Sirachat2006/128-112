```python
# โดยให้ทำการคำนวณหาราคารวมของรหัสคำสั่งซื้อแต่ละรายการ โดยให้คำนวณ vat 7% และราคารวม vat สุทธิ
# เก็บข้อมูล รหัสคำสั่งซื้ (id) ราคารวม vat และราคารวม vat ลงใน list ของ tuple
# จากนั้นแสดงรายงานยอดขายทุก tuple และยอดขายรวมทั้งหมด

#qty = จำนวน

orders = [
  {"id": 2001,"customer": "Anan", "city": "Korat","product":"Keyboard", "qty": 2, "price": 450},
  {"id": 2002,"customer": "Benja", "city": "Bangkok","product":"Mouse", "qty": 5,"price": 180},
  {"id": 2003,"customer": "Chai", "city": "Bangkok","product":"Monitor",  "qty": 1,"price": 3200},
  {"id": 2004,"customer": "Duang", "city": "Phuket","product":"Mouse",    "qty": 3,"price": 180},
  {"id": 2005,"customer": "Anan", "city": "Korat","product":"Monitor",  "qty": 2,"price": 3200},
  {"id": 2006,"customer": "Ekk",  "city": "Bangkok","product":"Keyboard", "qty": 4, "price": 450},
]

tuple = []
vat_rate = 0.07
total_sale = 0.0

for o in orders:
  total = o["qty"] * o["price"]
  vat = total * vat_rate
  total_vat = total + vat
  total_sale += total_vat
  tuple.append((o["id"], total, vat, total_vat, total_sale))

print("====================รายงานยอดขาย====================")
for tuple in tuple:
  print(f"ID {tuple[0]:<8} Total{tuple[1]:>5} Vat {tuple[2]:<5,.2f} Total Vat {tuple[3]:<12,.2f}")

print(f"============ยอดขายรวมทั้งหมด {total_sale:,.2f} "  "บาท============")
```
